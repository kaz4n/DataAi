```python
# Comprehensive column cleaning function
def clean_columns(df, null_threshold=40):
    """
    Clean DataFrame columns by removing:
    1. Completely null columns
    2. Single value columns
    3. Columns with null values above threshold%
    """
    columns_to_drop = []
    
    # 1. Find completely null columns
    completely_null = df.columns[df.isnull().all()].tolist()
    columns_to_drop.extend(completely_null)
    
    # 2. Find single value columns
    for column in df.columns:
        if df[column].nunique(dropna=True) == 1:
            columns_to_drop.append(column)
    
    # 3. Find columns with high null percentage
    null_percentages = (df.isnull().sum() / len(df)) * 100
    high_null_cols = df.columns[null_percentages > null_threshold].tolist()
    columns_to_drop.extend(high_null_cols)
    
    # Remove duplicates
    columns_to_drop = list(set(columns_to_drop))
    
    # Print summary
    print("Column Cleaning Summary:")
    print("-" * 50)
    print(f"Original columns: {len(df.columns)}")
    print(f"Columns to be dropped: {len(columns_to_drop)}")
    print("\nColumns being dropped:")
    for col in columns_to_drop:
        reason = []
        if col in completely_null:
            reason.append("completely null")
        if df[col].nunique(dropna=True) == 1:
            reason.append("single value")
        if col in high_null_cols:
            reason.append(f"{null_percentages[col]:.1f}% null")
        print(f"- {col}: {', '.join(reason)}")
    
    # Create clean dataset
    df_clean = df.drop(columns=columns_to_drop)
    
    print(f"\nRemaining columns: {len(df_clean.columns)}")
    return df_clean

# Apply the cleaning
data_clean = clean_columns(data, null_threshold=40)

# Optional: Update original dataframe
# data = data_clean

# Verify cleaning results
print("\nVerification:")
print("-" * 50)
print("Remaining column names:")
print(data_clean.columns.tolist())
```

```python
# Identify null columns and their null percentages
def analyze_null_columns(df, threshold=1.0):
    """
    Analyze null values in DataFrame columns
    threshold: proportion of nulls required to be considered a "null column" (1.0 = 100% null)
    """
    # Calculate null percentages for each column
    null_percentages = (df.isnull().sum() / len(df)) * 100
    
    # Find completely null columns (100% null)
    completely_null = df.columns[df.isnull().all()].tolist()
    
    # Find columns with null percentage above threshold
    high_null_cols = df.columns[df.isnull().mean() > threshold].tolist()
    
    # Print results
    print("Null Analysis Results:")
    print("-" * 50)
    
    print("\nCompletely null columns:")
    print(f"Found {len(completely_null)} completely null columns:")
    for col in completely_null:
        print(f"- {col}")
    
    print("\nNull percentages for all columns:")
    print("-" * 50)
    for col in df.columns:
        pct = null_percentages[col]
        count = df[col].isnull().sum()
        print(f"{col}: {pct:.2f}% null ({count} null values)")
    
    return completely_null, null_percentages

# Run the analysis
completely_null_columns, null_percentages = analyze_null_columns(data_clean)
```
