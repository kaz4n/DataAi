```python
import pandas as pd
import numpy as np
# Load data
df = pd.read_csv("E:\\Codding work\\\DataForDataScienceAndAI\\Cars_data.csv")
```

```python
# Loop to show unique values, null counts, null percentage, and categorize columns
numerical_cols = []
categorical_cols = []

for col in df.columns:
    unique_vals = df[col].unique()
    num_null = df[col].isnull().sum()
    pct_null = 100 * num_null / len(df)
    dtype = df[col].dtype

    print(f"Column: {col}")
    print(f"  Unique values: {unique_vals[:10]}{' ...' if len(unique_vals) > 10 else ''}")
    print(f"  Null values: {num_null} ({pct_null:.2f}%)")
    print(f"  Dtype: {dtype}")

    if pd.api.types.is_numeric_dtype(df[col]):
        numerical_cols.append(col)
        print("  Category: Numerical")
    else:
        categorical_cols.append(col)
        print("  Category: Categorical")
    print("-" * 40)

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)
```
